```python
import numpy as np
import pandas as pd
```

```python
credits = pd.read_csv("data/tmdb_5000_credits.csv")
movies = pd.read_csv("data/tmdb_5000_movies.csv")
```

```python
movies.head(1)
```

```python
credits.head(1)
```

```python
movies = movies.merge(credits, on="title")
movies.head(1)
```

```python
movies = movies[["id", "title", "genres", "overview", "cast", "crew", "keywords"]]
```

```python
movies.head()
```

```python
movies.info()
```

```python
movies.isnull().sum()
```

```python
movies = movies.dropna()
```

```python
movies.duplicated().sum()
```

```python
movies.head(1)
```

```python
movies.iloc[0].genres
```

```python
import ast


def convert(obj):
    List = []
    for i in ast.literal_eval(obj):
        List.append(i["name"])
    return List


def fetch_top_three_cast(obj):
    List = []
    counter = 0
    for i in ast.literal_eval(obj):
        if convert != 3:
            List.append(i["name"])
            counter += 1
        else:
            break
    return List


def fetch_director(obj):
    List = []
    for i in ast.literal_eval(obj):
        if i["job"] == "Director":
            List.append(i["name"])
            break
    return List
```

```python
movies["genres"] = movies["genres"].apply(convert)
```

```python
movies.iloc[0].keywords
```

```python
movies["keywords"] = movies["keywords"].apply(convert)
```

```python
movies.head(1)
```

```python
movies["cast"] = movies["cast"].apply(fetch_top_three_cast)
```

```python
movies["crew"] = movies["crew"].apply(fetch_director)
```

```python
movies.head(1)
```

```python
movies["overview"] = movies["overview"].apply(lambda x: x.split())
```

```python
# removing spaces between cast, crew, and keyword names

movies["genres"] = movies["genres"].apply(lambda x: [i.replace(" ", "") for i in x])
movies["keywords"] = movies["keywords"].apply(lambda x: [i.replace(" ", "") for i in x])
movies["crew"] = movies["crew"].apply(lambda x: [i.replace(" ", "") for i in x])
movies["cast"] = movies["cast"].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
movies.head(1)
```

```python
movies["tags"] = (
    movies["overview"]
    + movies["genres"]
    + movies["keywords"]
    + movies["cast"]
    + movies["crew"]
)
```

```python
movies.head(1)
```

```python
final_df = movies[["id", "title", "tags"]]
```

```python
final_df.head()
```

```python
final_df["tags"] = final_df["tags"].apply(lambda x: " ".join(x))
```

```python
final_df["tags"] = final_df["tags"].apply(lambda x: x.lower())
```

```python
final_df.head()
```

```python
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
def stem(txt):
    y = []
    for i in txt.split():
        y.append(ps.stem(i))
    return " ".join(y)

final_df["tags"] = final_df["tags"].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words="english")
```

```python
vectors = cv.fit_transform(final_df["tags"]).toarray()
```

```python
vectors[0]
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
def recommend(movie):
    movie_index = final_df[final_df["title"] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]

    for i in movies_list:
        print(final_df.iloc[i[0]].title)
```

```python
recommend("Batman")
```
